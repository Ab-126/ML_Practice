Create a 1D array containing numbers from 0 to 11 (inclusive), and reshape it into a $3 \times 4$ matrix. Then, extract all elements from the second row.


```python
# create and reshape
arr = np.arange(12).reshape(3, 4)
print("Matrix:\n", arr)

# Extract the second row
second_row = arr[1, ]
print("Second row:", second_row)
```

Given a random 1D array of 10 integers between 1 and 50, write a command to replace all even numbers in the array with -1

```python
np.random.seed(2)
arr = np.random.randint(1, 51, size=10)
print("Original:", arr)

# replace even numbers with -1
arr[arr % 2 == 0] = -1
print("Modified:", arr)
```

Given a $3 \times 3$ matrix filled with zeros, use broadcasting to add the array [1, 2, 3] to each row of the matrix

```python
matrix = np.zeros((3, 3), dtype=int)
print(matrix, "\n")
vector = np.array([1, 2, 3])

# Broadcasting adds the 1D vector to each row automatically
result = matrix + vector
print(result)
```

Create a $4 \times 4$ matrix with random integers between 1 and 100. Find:
- The maximum value in the entire matrix.
- The sum of each column.
- The mean value of each row.

```python
np.random.seed(2)
# arr = np.random.randint(1, 101, size=16).reshape(4, 4)
arr = np.random.randint(1, 101, size=(4, 4))
print(arr, "\n")

print("Max value: ", np.max(arr))
print("Columns sum", np.sum(arr, axis=0))
print("Row mean", np.mean(arr, axis=1))
```

Given an array of numbers, create a new array where any value less than 10 is replaced by 0, and any value greater than or equal to 10 is kept as is

```python
arr = np.array([5, 12, 3, 19, 8, 25])
print("original", arr)

# arr[arr < 10] = 0
res = np.where(arr < 10, 0, arr)
print("Modified", res)
```

Given a $4 \times 4$ matrix, write a command to extract rows at indices [0, 2] and columns at indices [1, 3] simultaneously.

```python
np.random.seed(42)
matrix = np.random.randint(1, 20, size=(4, 4))
print(matrix, "\n")

rows = np.array([0, 2])
cols = np.array([1, 3])

res = matrix[rows[:, np.newaxis], cols]
print("Extracted elements:\n", res)
```

Create two separate 1D arrays of size 4 containing random integers. Concatenate them horizontally (column-wise) and vertically (row-wise)

```python
np.random.seed(2)
arr1 = np.random.randint(0, 100, size=4)
arr2 = np.random.randint(0, 100, size=4)

v_stacked = np.vstack((arr1, arr2))
print("Vertical Stack:\n", v_stacked)

h_stacked = np.hstack((arr1, arr2))
print("Horizontal Stack ", h_stacked)
```

Given an array containing duplicate integer values, find all unique elements and count how many times each unique element appears.

```python
arr = np.array([10, 20, 10, 30, 20, 10, 40, 50, 30])

unique_vals, counts = np.unique(arr, return_counts=True)
print("Unique values: ", unique_vals)
print("Counts: ", counts)
```

Create two matrices: Matrix $A$ of shape $2 \times 3$ and Matrix $B$ of shape $3 \times 2$. Perform matrix multiplication to get a $2 \times 2$ result matrix.

```python
np.random.seed(2)
mat1 = np.random.randint(1, 100, size=(2, 3))
mat2 = np.random.randint(1, 100, size=(3, 2))

res = mat1 @ mat2
print(res)
```

Given an array containing NaN (Not a Number) or inf values, write a command to find the indices of all non-finite elements.

```python
arr = np.array([1.5, np.nan, 3.2, np.inf, 5.0, -np.inf])

# Find indices where values are not finite (NaN or inf)
non_finite_indices = np.where(~np.isfinite(arr))

print("Original array:", arr)
print("Indices of non-finite elements:", non_finite_indices)
```
